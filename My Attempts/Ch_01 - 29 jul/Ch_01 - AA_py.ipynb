## Calculate the Pupulation Mean and Populaiton Standard Deviation ##

Now that we have the data, we can calculate the **Population Mean** of the number of apples for sale at each store with the `mean()`function, which is part of the `statistics`module.

**Note:** The equation for the **Population Mean** is the same as the equation for the **Estimated Mean**, the only differnece is that for the **Population Mean** we use data from the entire population and for the **Estimated Mean** we only use a subset of the data.

```python
## Calculate the mean of the number of apples for sale
## because we are using the data from every single store
## we are calculating the Population Mean.
## Anyway, we'll save the value in a variable called pop_mean

pop_mean = statistics.mean(spend_n_save_df.num_apples)

## print out the population mean
pop_mean
```

Since the **Population Mean** has so many digits after the decimal point, it's quite a mouthful. Try rounding the number instead, using the function `round()`.

```python
round(pop_mean, ndigits=1)
```

Now let's calculate the  **Population Standard Deviation**. We can do that with the `pstdev()`function, which is also part of the `statistics`module.

```python
## calculate the population standard deviation and save it
## in a variable called pop_sd
pop_sd = statistics.pstdev(spend_n_save_df.num_apples)

## print out the population standard deviation
pop_sd
```

```python
pop_sum_man = sum((spend_n_save_df.num_apples - pop_mean) ** 2)
pop_var_man = pop_sum_man / spend_n_save_df.length

print("Population Variance, manual:", round(pop_var_man, ndigits=1), "\n")

pop_sd_man = pop_var_man ** 0.5

print("Population standard deviation, automatic:", 
      round(pop_sd, ndigits=1),
      "\n",
      "Population standard deviation, manual:",
      round(pop_sd_man, ndigits=1))
```

## Calculate the Estimated Mean and the Estimated Standard Deviation with a randomly selected subset of the data ##

Now that we know how to calculate the **Population Mean** and the **Population Standard Deviation**, let's learn how to calculate an **Estimated Mean** and an **Estimated Standard Deviation**.

The first thing we'll do is randomly select 5 values from the dataset. Specifically, we want to randomly select 5 values representing the number of apples for sale at a store.

```python
## Set seed for reproducibiility
np.random.seed(42)

## Select a random sample of size 5 without replacement
rand_sample = spend_n_save_df.num_apples.sample(n=5)

## print out the 5 values in rand_sample
rand_sample
```

Now, let's calculate the estiamted mean, just like before.

```python
## calculate the estimated mean
estimated_mean = statistics.mean(rand_sample)

## print out the estimated mean
estimated_mean
```

To calculate the estimated standard deviation, we can use the `stdev()`function that is part of the `statistics` module

```python
## Now calculate the estimated standard deviation
estimated_sd = statistics.stdev(rand_sample)
round(estimated_sd, ndigits=1)
```

Now let's get a sense for why the **Population Standard Deviation** is calculated with a slightly different formula than the **Estimated Standard Deviation**. First, remember the equation for the **Population Standard Deviation**. Let's compare it to the formula for the **Estimated Standard Deviation**.

\begin{array}{cc}
\text{Population Standard Deviation}
&
\text{Estimated Standard Deviation}
\\[1em]
\displaystyle \sqrt{\frac{\sum (x-\mu)^2}{N}}
&
\displaystyle \sqrt{\frac{\sum (x-\bar{x})^2}{n-1}}
\end{array}

Other than replacing the **Population Mean**, $\mu$, with the **Estimated Mean** $\bar{x}$, the big difference is the denominator. The **Population Standard Deviation** has $N$, the total number of measurements in the population, and the **Estimated Standard Deviation** has $n-1$, where $n$, is the number of measurements in the sample.

The reason the denominators are different is, int theory, because dividing by $n$ is the **Estimated Standard Deviation** will, on averafe, result in an underestimate of the **Population Standard Deviation**, and dividing by $n-1$ corrects for this bias.

So, let's see if this theory is actually true. We'll do this by selecting a bunch of samples, each with 5 randomly selected measurements and calculate the Standard Deviation both ways - using $n$ in the denominator and using $n-1$ in the denominator.

```python
# Set seed for reproducibility
np.random.seed(42)

max_samples = 1000 # this is how many times we'll get a sample of random values.

## we'll save all of the estimated standard deviations in these vectors.
estimated_sds_with_n_minus_1 = [0] * max_samples
estimated_sds_with_n         = [0] * max_samples

## Here's a loop where we'll randomly select a sample
## of 5 measurements and then calculate the SD two ways
for i in range(max_samples):

    ## get a sample of random values...
    rand_sample = spend_n_save_df.num_apples.sample(n=5)

    ## SD with n-1 (estimated)
    estimated_sds_with_n_minus_1[i] = statistics.stdev(rand_sample)

    ## SD witn n
    estimated_sds_with_n[i] = statistics.pstdev(rand_sample)
```

Now let's calculate and print the average SD that we just calculated for both equations...

```python
print("Average SD with n-1:", round(sum(estimated_sds_with_n_minus_1) / max_samples, 1))
print("Average SD with n:", round(sum(estimated_sds_with_n) / max_samples, 1))
```

...and compare those values to the actual **Population Standard Deviation**...

```python
round(pop_sd,1)
```

...and we see that, on average, when we have $n$ in the denominator, we underestimate the **Population Standard Deviation** more, than when we have $n-1$ in the denominator.
